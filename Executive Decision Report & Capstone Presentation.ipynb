```python
# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from IPython.display import display

plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["axes.titlesize"] = 16
plt.rcParams["axes.labelsize"] = 12
plt.rcParams["figure.dpi"] = 100
```

```python
# Load Dataset

file_path = "Exectudive decision report.csv"

df = pd.read_csv(file_path)

print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

display(df.head())
```

```python
# Data Overview

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nData Types:")
display(df.dtypes)

print("\nMissing Values:")
display(df.isnull().sum().sort_values(ascending=False))

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Prepare Data

data = df.copy()

# Convert date columns
data["signup_date"] = pd.to_datetime(data["signup_date"], errors="coerce")
data["last_purchase_date"] = pd.to_datetime(data["last_purchase_date"], errors="coerce")

# Convert numeric columns
numeric_columns = [
    "total_spent",
    "avg_order_value",
    "marketing_spend_per_user",
    "churn",
    "lifetime_value",
    "satisfaction_score",
    "nps_score"
]

for col in numeric_columns:
    data[col] = pd.to_numeric(data[col], errors="coerce")

print("Data preparation completed.")
print("Final Shape:", data.shape)
```

```python
# Executive KPIs

total_revenue = data["total_spent"].sum()
average_order_value = data["avg_order_value"].mean()
customer_acquisition_cost = data["marketing_spend_per_user"].mean()
churn_rate = data["churn"].mean() * 100
average_lifetime_value = data["lifetime_value"].mean()

total_customers = len(data)

print("EXECUTIVE KPI SUMMARY")
print("-" * 35)

print(f"Total Revenue: ₹{total_revenue:,.2f}")
print(f"Average Order Value: ₹{average_order_value:,.2f}")
print(f"Customer Acquisition Cost: ₹{customer_acquisition_cost:,.2f}")
print(f"Churn Rate: {churn_rate:.2f}%")
print(f"Average Lifetime Value: ₹{average_lifetime_value:,.2f}")
print(f"Total Customers: {total_customers:,}")
```

```python
# Revenue by Acquisition Channel

channel_revenue = (
    data.groupby("acquisition_channel")["total_spent"]
    .sum()
    .sort_values(ascending=False)
)

display(channel_revenue)
```

```python
# Revenue by Acquisition Channel Graph

plt.figure(figsize=(10, 6))

channel_revenue.plot(kind="bar")

plt.title("Revenue by Acquisition Channel")
plt.xlabel("Acquisition Channel")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# Revenue by Acquisition Channel Graph

plt.figure(figsize=(10, 6))

channel_revenue.plot(kind="bar")

plt.title("Revenue by Acquisition Channel")
plt.xlabel("Acquisition Channel")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# Customer Acquisition Cost by Channel

channel_cac = (
    data.groupby("acquisition_channel")["marketing_spend_per_user"]
    .mean()
    .sort_values(ascending=False)
)

display(channel_cac)
```

```python
# CAC by Acquisition Channel Graph

plt.figure(figsize=(10, 6))

channel_cac.plot(kind="bar")

plt.title("Average Customer Acquisition Cost by Channel")
plt.xlabel("Acquisition Channel")
plt.ylabel("CAC (₹)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# Revenue Trend

monthly_revenue = (
    data.dropna(subset=["last_purchase_date"])
    .set_index("last_purchase_date")
    .resample("ME")["total_spent"]
    .sum()
)

display(monthly_revenue.head())
```

```python
# Revenue Trend Graph

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_revenue.index,
    monthly_revenue.values,
    marker="o",
    linewidth=2
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Revenue by Country

country_revenue = (
    data.groupby("country")["total_spent"]
    .sum()
    .sort_values(ascending=False)
)

display(country_revenue.head(10))
```

```python
# Top 10 Countries by Revenue

top_countries = country_revenue.head(10).sort_values()

plt.figure(figsize=(10, 6))

top_countries.plot(kind="barh")

plt.title("Top 10 Countries by Revenue")
plt.xlabel("Revenue (₹)")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
# Customer Lifetime Value Analysis

print("Average Lifetime Value:", f"₹{data['lifetime_value'].mean():,.2f}")
print("Median Lifetime Value:", f"₹{data['lifetime_value'].median():,.2f}")
print("Maximum Lifetime Value:", f"₹{data['lifetime_value'].max():,.2f}")
```

```python
# Top Customers by Revenue

customer_columns = [
    "customer_id",
    "total_spent",
    "lifetime_value",
    "avg_order_value",
    "churn"
]

top_customers = (
    data[customer_columns]
    .sort_values("total_spent", ascending=False)
    .head(10)
)

display(top_customers)
```

```python
# Churn Analysis

churn_summary = data["churn"].value_counts()

print("Churn Counts:")
display(churn_summary)

print(f"\nOverall Churn Rate: {churn_rate:.2f}%")
```

```python
# Churn Graph

churn_labels = ["Active", "Churned"]
churn_values = [
    (data["churn"] == 0).sum(),
    (data["churn"] == 1).sum()
]

plt.figure(figsize=(8, 6))

plt.bar(churn_labels, churn_values)

plt.title("Customer Churn Distribution")
plt.xlabel("Customer Status")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

```python
# Churn by Acquisition Channel

churn_by_channel = (
    data.groupby("acquisition_channel")["churn"]
    .mean()
    .sort_values(ascending=False) * 100
)

display(churn_by_channel)
```

```python
# Churn Rate by Acquisition Channel Graph

plt.figure(figsize=(10, 6))

churn_by_channel.plot(kind="bar")

plt.title("Churn Rate by Acquisition Channel")
plt.xlabel("Acquisition Channel")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Customer Satisfaction and NPS

average_satisfaction = data["satisfaction_score"].mean()
average_nps = data["nps_score"].mean()

print(f"Average Satisfaction Score: {average_satisfaction:.2f}")
print(f"Average NPS Score: {average_nps:.2f}")
```

```python
# Churn vs Satisfaction

satisfaction_churn = (
    data.groupby("churn")["satisfaction_score"]
    .mean()
)

display(satisfaction_churn)
```

```python
# Churn vs Satisfaction

satisfaction_churn = (
    data.groupby("churn")["satisfaction_score"]
    .mean()
)

display(satisfaction_churn)
```

```python
# Satisfaction by Customer Status

labels = ["Active", "Churned"]
values = [
    satisfaction_churn.get(0, 0),
    satisfaction_churn.get(1, 0)
]

plt.figure(figsize=(8, 6))

plt.bar(labels, values)

plt.title("Average Satisfaction: Active vs Churned Customers")
plt.xlabel("Customer Status")
plt.ylabel("Average Satisfaction Score")

plt.tight_layout()
plt.show()
```

```python
# Marketing Efficiency

marketing_summary = (
    data.groupby("acquisition_channel")
    .agg(
        Revenue=("total_spent", "sum"),
        Marketing_Spend=("marketing_spend_per_user", "sum"),
        Customers=("customer_id", "count")
    )
)

marketing_summary["Revenue_per_Marketing_Rupee"] = (
    marketing_summary["Revenue"] /
    marketing_summary["Marketing_Spend"]
)

display(
    marketing_summary.sort_values(
        "Revenue_per_Marketing_Rupee",
        ascending=False
    )
)
```

```python
# Marketing Efficiency Graph

efficiency = marketing_summary["Revenue_per_Marketing_Rupee"].sort_values()

plt.figure(figsize=(10, 6))

efficiency.plot(kind="barh")

plt.title("Revenue Generated per ₹1 of Marketing Spend")
plt.xlabel("Revenue / Marketing Spend")
plt.ylabel("Acquisition Channel")

plt.tight_layout()
plt.show()
```

```python
# Revenue by Subscription Type

subscription_revenue = (
    data.groupby("subscription_type")["total_spent"]
    .sum()
    .sort_values(ascending=False)
)

display(subscription_revenue)
```

```python
# Revenue by Subscription Type Graph

plt.figure(figsize=(10, 6))

subscription_revenue.plot(kind="bar")

plt.title("Revenue by Subscription Type")
plt.xlabel("Subscription Type")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
# Premium Customer Analysis

if "is_premium_user" in data.columns:

    premium_revenue = (
        data.groupby("is_premium_user")["total_spent"]
        .agg(["count", "sum", "mean"])
    )

    display(premium_revenue)

else:
    print("is_premium_user column not found.")
```

```python
# High-Risk Customer Analysis

risk_columns = [
    "customer_id",
    "total_spent",
    "lifetime_value",
    "avg_order_value",
    "churn",
    "satisfaction_score"
]

risk_customers = data[risk_columns].copy()

risk_customers["Risk_Score"] = (
    risk_customers["churn"] * 2
    + (risk_customers["satisfaction_score"] < 3).astype(int)
)

risk_customers = risk_customers.sort_values(
    "Risk_Score",
    ascending=False
)

display(risk_customers.head(10))
```

```python
# Refund Analysis

if "refund_requested" in data.columns:

    refund_rate = data["refund_requested"].mean() * 100

    print(f"Refund Request Rate: {refund_rate:.2f}%")

    refund_summary = data["refund_requested"].value_counts()

    display(refund_summary)

else:
    print("refund_requested column not found.")
```

```python
# Current Customer Value

current_ltv = data["lifetime_value"].mean()

churned_customers = (data["churn"] == 1).sum()

print(f"Average Customer Lifetime Value: ₹{current_ltv:,.2f}")
print(f"Current Churned Customers: {churned_customers:,}")
```

```python
# ROI Projection

churn_reduction_target = 0.10

customers_saved = churned_customers * churn_reduction_target

estimated_value_saved = customers_saved * current_ltv

retention_investment = data["marketing_spend_per_user"].mean() * customers_saved

estimated_roi = (
    (estimated_value_saved - retention_investment)
    / retention_investment
) * 100 if retention_investment > 0 else 0

print("ROI PROJECTION")
print("-" * 30)

print(f"Target churn reduction: {churn_reduction_target * 100:.0f}%")
print(f"Estimated customers retained: {customers_saved:,.0f}")
print(f"Estimated value protected: ₹{estimated_value_saved:,.2f}")
print(f"Estimated retention investment: ₹{retention_investment:,.2f}")
print(f"Estimated ROI: {estimated_roi:.2f}%")
```

```python
# Executive Findings

top_channel = channel_revenue.index[0]
top_channel_revenue = channel_revenue.iloc[0]

highest_churn_channel = churn_by_channel.index[0]
highest_churn_value = churn_by_channel.iloc[0]

best_efficiency_channel = (
    marketing_summary["Revenue_per_Marketing_Rupee"]
    .idxmax()
)

top_country = country_revenue.index[0]
top_country_revenue = country_revenue.iloc[0]

print("EXECUTIVE FINDINGS")
print("=" * 40)

print(
    f"1. {top_channel} is the leading revenue acquisition channel "
    f"with ₹{top_channel_revenue:,.2f} in revenue."
)

print(
    f"2. {highest_churn_channel} has the highest churn rate "
    f"at {highest_churn_value:.2f}%."
)

print(
    f"3. {best_efficiency_channel} generates the highest "
    f"revenue relative to marketing spend."
)

print(
    f"4. {top_country} is the highest-revenue country "
    f"with ₹{top_country_revenue:,.2f}."
)

print(
    f"5. Overall customer churn is {churn_rate:.2f}%."
)

print(
    f"6. Average customer lifetime value is "
    f"₹{average_lifetime_value:,.2f}."
)
```

```python
# Strategic Recommendations

print("STRATEGIC RECOMMENDATIONS")
print("=" * 40)

print("""
1. OPTIMIZE ACQUISITION SPENDING
Focus marketing investment on acquisition channels that generate
strong revenue relative to customer acquisition cost.

2. REDUCE CUSTOMER CHURN
Prioritize retention campaigns for high-churn acquisition channels
and customers showing lower satisfaction or engagement.

3. INCREASE CUSTOMER LIFETIME VALUE
Use personalized offers, premium services and targeted engagement
to increase average order value and customer lifetime value.
""")
```

```python
# Final KPI Table

kpi_table = pd.DataFrame({
    "KPI": [
        "Total Revenue",
        "Average Order Value",
        "Customer Acquisition Cost",
        "Churn Rate",
        "Average Lifetime Value",
        "Average Satisfaction",
        "Average NPS"
    ],
    "Value": [
        f"₹{total_revenue:,.2f}",
        f"₹{average_order_value:,.2f}",
        f"₹{customer_acquisition_cost:,.2f}",
        f"{churn_rate:.2f}%",
        f"₹{average_lifetime_value:,.2f}",
        f"{average_satisfaction:.2f}",
        f"{average_nps:.2f}"
    ]
})

display(kpi_table)
```

```python
# Save Analysis Results

kpi_table.to_csv(
    "Executive_KPI_Summary.csv",
    index=False
)

marketing_summary.to_csv(
    "Marketing_Efficiency_Analysis.csv"
)

country_revenue.to_csv(
    "Country_Revenue_Analysis.csv"
)

channel_revenue.to_csv(
    "Channel_Revenue_Analysis.csv"
)

churn_by_channel.to_csv(
    "Churn_By_Channel.csv"
)

print("Analysis files saved successfully.")
```
